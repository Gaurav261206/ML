10 : Solve a simple grid world problem using reinforcement earning, where an agent learns to reach a target location while avoiding obstacles.

```python
import numpy as np
import random

grid = []
for i in range(3):
    row = []
    for j in range(3):
        row.append(".")
    grid.append(row)

r = int(input("Enter the row for start(0-2) : "))
c = int(input("Enter the column for start(0-2) : "))
start = (r, c)
grid[r][c] = "S"

r = int(input("Enter the row for goal(0-2) : "))
c = int(input("Enter the column for goal(0-2) : "))
goal = (r, c)
grid[r][c] = "G"

r = int(input("Enter the row for obstacle(0-2) : "))
c = int(input("Enter the column for obstacle(0-2) : "))
obstacle = (r, c)
grid[r][c] = "X"

print("\nGrid World :")
for row in grid:
    print(*row)

Q = np.zeros((3, 3, 4))
actions = [(-1,0), (1, 0), (0, -1), (0, 1)]

alpha = 0.1
gamma = 0.9
epsilon = 0.2

for episodes in range(2000) :
    position = start

    for steps in range(50) :
        r, c = position

        if random.random() < epsilon :
            action = random.randint(0, 3)
        else :
            action = np.argmax(Q[r, c])

        dr, dc = actions[action]
        nr, nc = r+dr, c+dc

        reward = -1
        next_position = position

        if not (0 <= nr < 3 and 0 <= nc < 3) :
            reward = -5
        elif (nr, nc) == obstacle :
            reward = -10
        else :
            next_position = (nr, nc)

        if next_position == goal :
            reward = 10

        Q[r, c, action] += alpha * (
            reward + gamma * np.max(Q[next_position])
           - Q[r, c, action]
        )

        position = next_position
        if position == goal :
            break

position = start
path = [position]

for step in range(20) :
    if position == goal :
        break
    r,c = position
    action = np.argmax(Q[r, c])

    dr, dc = actions[action]
    next_position = (r+dr, c+dc)

    if not (0 <= next_position[0] < 3 and 0 <= next_position[1] < 3) :
        break

    if next_position == obstacle :
        break

    position = next_position
    path.append(position)

print("\nLearned Path :", path)

if position == goal :
    print("Goal Reached Successfully !!")
else :
    print("Path Not Found !!")
```
