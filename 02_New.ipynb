02 . Develop a Ridge and Lasso regression model to predict the number of bike rentals based on weather conditions and time.  

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import *
```

```python
df = pd.read_csv('bike.csv')
print("Dataset Shape :", df.shape)
df.sample(5)
```

```python
if df.isnull().sum().sum() > 0:
    print("Missing values are present.")
    print(df.isnull().sum())
else:
    print("No missing values found.")
```

```python
df.columns
```

```python
x = df.drop(['dteday', 'cnt', 'casual', 'registered'], axis=1)
y = df['cnt']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
print("Model Training Completed !!")

lr_pred = lr.predict(x_test)
print("Prediction :", lr_pred)
```

```python
rg = Ridge(alpha=0.1)
rg.fit(x_train, y_train)
print("Model Training Completed !!")

rg_pred = rg.predict(x_test)
print("Prediction :", rg_pred)
```

```python
ls = Lasso(alpha=0.1)
ls.fit(x_train, y_train)
print("Model Training Completed !!")

ls_pred = ls.predict(x_test)
print("Prediction :", ls_pred)
```

```python
print("\n===== Linear Regression =====")
print("R2 Score :", r2_score(y_test, lr_pred))
print("MSE :", mean_squared_error(y_test, lr_pred))

print("\n===== Ridge Regression =====")
print("R2 Score :", r2_score(y_test, rg_pred))
print("MSE      :", mean_squared_error(y_test, rg_pred))

print("\n===== Lasso Regression =====")
print("R2 Score :", r2_score(y_test, ls_pred))
print("MSE      :", mean_squared_error(y_test, ls_pred))
```
